# LangGraph Clinical Workflow Demo

Run `run_workflow` for all five demo patients and print each `graph_path`.

```python
import os
from pathlib import Path

ROOT = Path("..").resolve() if (Path("..") / "langgraph_flows").exists() else Path(".").resolve()
os.chdir(ROOT)
os.environ["USE_MOCK_LLM"] = "1"
```

```python
from langgraph_flows.clinical_workflow import run_workflow

DEMO_CASES = [
    ("PAC-001", "Protocolo para febre e tosse?"),
    ("PAC-002", "Qual conduta para dor abdominal?"),
    ("PAC-003", "Follow-up oncologia de mama?"),
    ("PAC-004", "Interação medicamentosa?"),
    ("PAC-005", "Consulta geral — orientações?"),
]
```

```python
for patient_id, query in DEMO_CASES:
    state = run_workflow(patient_id, query)
    path = " → ".join(state["graph_path"])
    alerts = len(state.get("alerts", []))
    print(f"{patient_id}: {path}  (alertas={alerts})")
```

```python
# PAC-002 detail: pending exam branch
pac002 = run_workflow("PAC-002", "Qual conduta para dor abdominal?")
print("Alertas:", pac002["alerts"])
print("Exames pendentes:", pac002["pending_exams"])
```
