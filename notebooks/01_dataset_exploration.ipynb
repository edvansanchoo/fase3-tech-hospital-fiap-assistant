# Dataset Exploration

Load processed JSONL splits from `data/processed/` and inspect record counts and samples.

```python
import json
from pathlib import Path

ROOT = Path("..").resolve() if (Path("..") / "data" / "processed").exists() else Path(".").resolve()
PROCESSED = ROOT / "data" / "processed"
SPLITS = ("train", "val", "test")
```

```python
def load_jsonl(path: Path) -> list[dict]:
    records = []
    if not path.exists():
        return records
    with path.open(encoding="utf-8") as handle:
        for line in handle:
            line = line.strip()
            if line:
                records.append(json.loads(line))
    return records

datasets = {split: load_jsonl(PROCESSED / f"{split}.jsonl") for split in SPLITS}
counts = {split: len(rows) for split, rows in datasets.items()}
print("Record counts:", counts)
print("Total:", sum(counts.values()))
```

```python
import pandas as pd

rows = []
for split, records in datasets.items():
    for record in records[:3]:
        rows.append(
            {
                "split": split,
                "instruction": record.get("instruction", ""),
                "input": record.get("input", ""),
                "output": record.get("output", "")[:120] + "..." if len(record.get("output", "")) > 120 else record.get("output", ""),
            }
        )

pd.DataFrame(rows)
```
